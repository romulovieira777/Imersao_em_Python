## Variáveis

Variáveis podem ser entendidas como uma caixa para armazenar informações. Em Python não existe um comando para declarar uma variável. Uma variável é criada no momento em que um valort é atribuido a ela:

```python
x = 5
y = "John"
```

```python
print(x)
print(y)
```

```python
x   # não vai exibir o x
y   # apenas a última linha é exibida no output da célula se o comando for apenas uma variável
```

```python
x
```

```python
y
```

Ao receber um novo valor, a variável descarta o valor que estava armazenado anteriormente. O novo valor não precisa ter o mesmo tipo.

```python
x = 4        # x armazena um valor do tipo inteiro
x = "Sally"  # é criada uma nova variável x, agora com valor string
print(x)
```

```python
# Python é case-sensitive
i = 4
I = 5

print(i)
print(I)
```

**Observações sobre nomes de variáveis**

Uma variável pode ter um nome simples (como "x" ou "y") ou pode ter um nome descritivo (idade, nome, volume_total). Mas existem algumas regras a serem seguidas:

* Um nome de variável deve começar com uma letra ou um unerline.
* Um nome de variável não pode começar com um número.
* Um nome de variável só pode conter caracteres alfa-numéricos e underlines (A-z, 0-9, e _ ).
* Nomes de variáveis são case-sensitive (idade, Idade e IDADE são três variáveis diferentes).

### Tipos de Dados

Em programação tipos de dados é um conceito muito importante. AS variáveis podem armazenar vários tipos diferentes de dados. E tipos diferentes de dasdos podem fazer coisas diferentes...

Python tem os seguintes tipos de dados:
   * Texto: ``str``
   * Números: ``int``, ``float``
   * Sequências: ``list``, ``tuple``, ``range``
   * Mapeamentos: ``dict``
   * Conjuntos: ``set``, ``frozenset``
   * Booleanos: ``bool``
