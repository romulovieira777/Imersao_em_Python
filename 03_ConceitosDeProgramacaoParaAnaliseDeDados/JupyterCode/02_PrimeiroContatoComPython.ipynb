# Intro

### O que é Python

Python é uma lingugem de programação. Ela foi criada por Guido Van Rossum e lançada em 1991. Ela é usada pra desenvolvimento de sistemas web, criação de software, matemática e scripting.

### Sintaxe

- Python foi desenhado para facilitar a leitura e tem algumas similaridades com a língua inglesa
- Utiliza a quebra de linha como o final de um comando, ao contrário da maioria das linguagens que , em geral utilizam parenteses ou ponto e vírgula.
- Identação faz parte da sintaxe. Os espaços em branco são utilizados para definir escopo de loops, funções e classes. Outras linguagens de programação normalmente utilizam chaves para esse propósito.

### Quick Start

Python é uma linguagem interpretada, isso significa que ao criar um arquivo .py em um editor de texto é possível colocar esse arquivo em um interpretador python e executá-lo.

Para executar um arquivo .py precisamos ter o python instalado na nossa máquina. Para verificar se ele já está instalado podemos rodar o seguinte comando em um terminal:

    C:\Users\Your Name>python --version

Se você não tiver o python instalado, basta ir até esse site e fazer o download de graça: https://www.python.org/

# A linha de comando

Python também permite que nós utilizemos a linha de comandos para rodar comandos e executar nosso código de forma mais rápida. Para fazer isso, primeiro digitamos o seguinte comando em um terminal:

  C:\Users\Your Name>python

Agora executaremos o mesmo código que utilizamos no arquivo helloworld.py:

  print("Hello World")

para finalizar a execução da linha de comando do python basta digitar a quelquer momento:

  exit()

Enquanto em várias linguagens de programação identação é usada apenas para facilitar a leitura, em Python ela é muito importante.

# Comments

Python também nos permite escrever comentários. Esses comantários podem ser utilizados para documentar nosso código, para evitar que uma linha de código seja executada ou por qualquer outra razão.

Comentários em Python começam com ``#``, tudo o que vem depois da cerquilha será tratado como comentário e ignorado pelo interpretador:

```python
# Um comentário.
print("Hello, World!")
```

```python
print("Hello, World!") # Isso vale?
```

```python
# E agora? O que será impresso
# print("Hello, World!")
print("Cheers, Mate!")
```

```python
'''
Esse é um comentário
longo
'''
print("Hello")
```
