```python
# Lendo o arquivo previamente com a opção inferSchema desligada
df = spark \
    .read.format("csv") \
    .option("inferSchema", "false") \
    .option("header", "true") \
    .csv(arquivo)
```

## Consultando os dados de um dataframe

```python
from pyspark.sql.functions import max
df.select(max("count")).take(1)
```

```python
# Filtrando linhas de um dataframe usando filter
df.filter("delay < 2").show(2)
```

```python
# Usando where (uma alias par o metodo filter)
df.where("delay < 2").show(2)
```

```python
# Ordena o dataframe pela coluna count em ordem decrescente
from pyspark.sql.functions import desc, asc, expr
df.orderBy(expr("delay desc")).show(10)
```

```python
# Visualizando estatísticas descritivas do dataframe
df.describe().show()
```

```python
# Iterando sobre todas as linhas do dataframe
for i in df.collect():
    print(i)
    print(i[0], i[1] * 2)
```

```python
# Adicionando uma coluna ao dataframe
df = df.withColumn("Nova Coluna", df["delay"] + 2)
df.show(10)
```

```python
# Removendo a coluna adicionada
df = df.drop("Nova Coluna")
df.show(10)
```

```python
# Renomeando uma coluna do dataframe
df = df.withColumnRenamed("Nova Coluna", "New Column").show()
```

```python
# Filtrando dados nullos
df.filter("delay is Null").show(10)
```

```python
# Filtrando valores missing
df.filter(df.delay.isNull()).show(10)
```

## Preenchendo Dados Missing

```python
# Preenchendo os dados missing com o valor 0
df.na.fill(value=0).show(10)
```

```python
# Preenche valores missing com valor 0 apenas da coluna delay
df.na.fill(value=0, subset=["delay"]).show(10)
```

```python
# Preenche os dados com valores de string vazia
df.na.fill(value="").show(10)
```

```python
# Remove qualquer linha nula de qualquer coluna
df.na.drop().show(10)
```
